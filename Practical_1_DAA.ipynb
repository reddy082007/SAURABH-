```python
import time

# Input from the user for Insertion Sort
n = int(input("Enter the number of elements for Insertion Sort: "))

arr = []
print(f"Enter {n} elements for Insertion Sort (space-separated):")
# Read all elements as a single string, split by space, and convert to int
input_elements_str = input()
arr = list(map(int, input_elements_str.split()))

# Start timer
start_time = time.time()

# Insertion Sort
for i in range(1, len(arr)):
    key = arr[i]
    j = i - 1

    while j >= 0 and arr[j] > key:
        arr[j + 1] = arr[j]
        j -= 1

    arr[j + 1] = key

# Stop timer
end_time = time.time()

# Calculate execution time
execution_time = end_time - start_time

# Display sorted array
print("\nSorted Array (Insertion Sort):")
print(arr)

# Display execution time
print(f"\nExecution Time (Insertion Sort): {execution_time:.6f} seconds")

# Display time complexity
print("\nTime Complexity (Insertion Sort):")
print("Best Case    : O(n)")
print("Average Case : O(n^2)")
print("Worst Case   : O(n^2)")
print("Space Complexity: O(1)")

# Merge Sort Function
def merge_sort(arr):
    if len(arr) > 1:
        mid = len(arr) // 2

        left = arr[:mid]
        right = arr[mid:]

        # Recursively sort both halves
        merge_sort(left)
        merge_sort(right)

        i = j = k = 0

        # Merge the two halves
        while i < len(left) and j < len(right):
            if left[i] < right[j]:
                arr[k] = left[i]
                i += 1
            else:
                arr[k] = right[j]
                j += 1
            k += 1

        # Copy remaining elements of left half
        while i < len(left):
            arr[k] = left[i]
            i += 1
            k += 1

        # Copy remaining elements of right half
        while j < len(right):
            arr[k] = right[j]
            j += 1
            k += 1

# User input for Merge Sort
n = int(input("Enter the number of elements for Merge Sort: "))

arr = []
print(f"Enter {n} elements for Merge Sort (space-separated):")
# Read all elements as a single string, split by space, and convert to int
input_elements_str = input()
arr = list(map(int, input_elements_str.split()))

# Start timer
start_time = time.time()

# Perform Merge Sort
merge_sort(arr)

# Stop timer
end_time = time.time()

# Calculate execution time
execution_time = end_time - start_time

# Display sorted array
print("\nSorted Array (Merge Sort):")
print(arr)

# Display execution time
print(f"\nExecution Time (Merge Sort): {execution_time:.6f} seconds")

# Display time complexity
print("\nTime Complexity (Merge Sort):")
print("Best Case    : O(n log n)")
print("Average Case : O(n log n)")
print("Worst Case   : O(n log n)")
print("Space Complexity: O(n)")

# Partition function
def partition(arr, low, high):
    pivot = arr[high]
    i = low - 1

    for j in range(low, high):
        if arr[j] <= pivot:
            i += 1
            arr[i], arr[j] = arr[j], arr[i]

    arr[i + 1], arr[high] = arr[high], arr[i + 1]
    return i + 1

# Quick Sort function
def quick_sort(arr, low, high):
    if low < high:
        pi = partition(arr, low, high)

        quick_sort(arr, low, pi - 1)
        quick_sort(arr, pi + 1, high)

# User input for Quick Sort
n = int(input("Enter the number of elements for Quick Sort: "))

arr = []
print(f"Enter {n} elements for Quick Sort (space-separated):")
input_elements_str = input()
arr = list(map(int, input_elements_str.split()))

# Start timer
start_time = time.time()

# Perform Quick Sort
quick_sort(arr, 0, len(arr) - 1)

# Stop timer
end_time = time.time()

# Calculate execution time
execution_time = end_time - start_time

# Display sorted array
print("\nSorted Array (Quick Sort):")
print(arr)

# Display execution time
print(f"\nExecution Time (Quick Sort): {execution_time:.6f} seconds")

# Display time complexity
print("\nTime Complexity (Quick Sort):")
print("Best Case    : O(n log n)")
print("Average Case : O(n log n)")
print("Worst Case   : O(n^2)")
print("Space Complexity: O(log n)")

# Input from the user for Bubble Sort 1
n = int(input("Enter the number of elements for Bubble Sort 1: "))

arr = []
print(f"Enter {n} elements for Bubble Sort 1 (space-separated):")
input_elements_str = input()
arr = list(map(int, input_elements_str.split()))

# Start timer
start_time = time.time()

# Bubble Sort
for i in range(n - 1):
    for j in range(n - i - 1):
        if arr[j] > arr[j + 1]:
            arr[j], arr[j + 1] = arr[j + 1], arr[j]

# Stop timer
end_time = time.time()

# Calculate execution time
execution_time = end_time - start_time

# Display sorted array
print("\nSorted Array (Bubble Sort):")
print(arr)

# Display execution time
print(f"\nExecution Time (Bubble Sort): {execution_time:.6f} seconds")

# Display time complexity
print("\nTime Complexity (Bubble Sort):")
print("Best Case    : O(n^2)")
print("Average Case : O(n^2)")
print("Worst Case   : O(n^2)")
print("Space Complexity: O(1)")

# Input from the user for Bubble Sort 2 (duplicate for demonstration)
n = int(input("Enter the number of elements for Bubble Sort 2: "))

arr = []
print(f"Enter {n} elements for Bubble Sort 2 (space-separated):")
input_elements_str = input()
arr = list(map(int, input_elements_str.split()))

# Start timer
start_time = time.time()

# Bubble Sort
for i in range(n - 1):
    for j in range(n - i - 1):
        if arr[j] > arr[j + 1]:
            arr[j], arr[j + 1] = arr[j + 1], arr[j]

# Stop timer
end_time = time.time()

# Calculate execution time
execution_time = end_time - start_time

# Display sorted array
print("\nSorted Array (Bubble Sort):")
print(arr)

# Display execution time
print(f"\nExecution Time (Bubble Sort): {execution_time:.6f} seconds")

# Display time complexity
print("\nTime Complexity (Bubble Sort):")
print("Best Case    : O(n^2)")
print("Average Case : O(n^2)")
print("Worst Case   : O(n^2)")
print("Space Complexity: O(1)")


# Input from the user for Selection Sort
n = int(input("Enter the number of elements for Selection Sort: "))

arr = []
print(f"Enter {n} elements for Selection Sort (space-separated):")
input_elements_str = input()
arr = list(map(int, input_elements_str.split()))

# Start timer
start_time = time.time()

# Selection Sort
for i in range(n - 1):
    min_index = i
    for j in range(i + 1, n):
        if arr[j] < arr[min_index]:
            min_index = j

    # Swap the minimum element with the current element
    arr[i], arr[min_index] = arr[min_index], arr[i]

# Stop timer
end_time = time.time()

# Calculate execution time
execution_time = end_time - start_time

# Display sorted array
print("\nSorted Array (Selection Sort):")
print(arr)

# Display execution time
print(f"\nExecution Time (Selection Sort): {execution_time:.6f} seconds")

# Display time complexity
print("\nTime Complexity (Selection Sort):")
print("Best Case    : O(n^2)")
print("Average Case : O(n^2)")
print("Worst Case   : O(n^2)")
print("Space Complexity: O(1)")
```
