```python
from collections import deque

graph = {
    'A': ['B', 'C'],
    'B': ['D', 'E'],
    'C': ['F'],
    'D': [],
    'E': ['F'],
    'F': []
}
print("graph:",graph)
```

# BFS

```python

def bfs(graph, start):
    visited = set()         
    queue = deque([start])  

    visited.add(start)

    while queue:
        node = queue.popleft()
        print(node, end=" ")

        for neighbor in graph[node]:
            if neighbor not in visited:
                visited.add(neighbor)
                queue.append(neighbor)

start_node = 'A'
print("BFS Traversal :")
bfs(graph, start_node)

print("graph:",graph)
```

# DFS

```python
def dfs(graph, node, visited):
    if node not in visited:
        print(node, end=" ")
        visited.add(node)

        for neighbor in graph[node]:
            dfs(graph, neighbor, visited)

visited = set()

print("DFS Traversal :")
dfs(graph, 'A', visited)
```
